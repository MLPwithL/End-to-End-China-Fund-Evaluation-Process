```python
import pandas as pd

# 读取 Barra 因子收益率
barra = pd.read_feather("BarraFactorReturn.txt")

# 日期处理
barra["time"] = pd.to_datetime(barra["time"].astype(str))
barra = barra.sort_values("time")

# 风格因子
style_factors = [
    "Beta",
    "BooktoPrice",
    "EarningYield",
    "Growth",
    "Leverage",
    "Liquidity",
    "Momentum",
    "NonlinearSize",
    "ResidualVolatility",
    "Size",
]

# 行业因子 = 除 time 和风格因子外的列
industry_factors = [
    c for c in barra.columns
    if c not in ["time"] + style_factors
]

# 分别保存
barra_style = barra[["time"] + style_factors]
barra_industry = barra[["time"] + industry_factors]

barra_style.reset_index(drop=True).to_feather("Barra风格因子收益率.feather")

barra_industry.reset_index(drop=True).to_feather("Barra行业因子收益率.feather")

# 合并总表保存
barra.reset_index(drop=True).to_feather("Barra因子收益率.feather")

print("风格因子:", barra_style.shape)
print("行业因子:", barra_industry.shape)
print("全部因子:", barra.shape)
```
