```python
import pandas as pd

nav = pd.read_feather("D:\BaiduNetdiskDownload\正则化基金数据\正则化基金数据\基金数据\开放式基金_NAV_clean.feather")
nav["time"] = pd.to_datetime(nav["time"].astype(str))
nav = nav[["time"]]

nav.reset_index(drop=True).to_feather("trading_days.feather")
trading_days = nav = pd.read_feather("trading_days.feather")
```

```python
import pandas as pd

trading_days = nav = pd.read_feather("trading_days.feather")

# 读取季度持仓日期
hold = pd.read_feather(
    "D:\BaiduNetdiskDownload\正则化基金数据\正则化基金数据\基金数据\开放式基金_StockHold_quarterly.feather"
)

hold_dates = (
    hold[["time"]]
    .drop_duplicates()
    .rename(columns={"time": "hold_date"})
    .sort_values("hold_date")
)

# 交易日 → 最近一期持仓日期
mapping = pd.merge_asof(
    trading_days,
    hold_dates,
    left_on="time",
    right_on="hold_date",
    direction="backward"
)

# 保存映射表
mapping.reset_index(drop=True).to_feather(
    "交易日_持仓日期映射.feather"
)

print(mapping.head())
```
